# Business Problem Understanding

This dataset contains various features related to cars, including the year of manufacture, selling price, kilometers driven, fuel type, seller type, transmission type, number of previous owners, mileage, and engine specifications. These attributes provide valuable insights into the factors influencing car prices and can be used to develop predictive models for estimating the selling price of cars.


- name: The name or model of the car (e.g., "Maruti Swift Dzire VDI").
- year: The year the car was manufactured (e.g., 2014).
- selling_price: The price at which the car is being sold (in your dataset, it's in Indian Rupees).
- km_driven: The total distance the car has traveled (in kilometers). A higher value means the car has been driven more.
- fuel: The type of fuel the car uses (e.g., Petrol, Diesel).
- seller_type: Indicates whether the car is being sold by an individual or a dealer.
- transmission: The type of transmission the car uses (e.g., Manual or Automatic).
- owner: The number of previous owners of the car (e.g., "First Owner" or "Second Owner").
- mileage(km/ltr/kg): The fuel efficiency of the car, showing how many kilometers it can travel per liter of fuel (higher means better mileage).
- engine: The engine capacity of the car, usually in cubic centimeters (cc). A higher value indicates a more powerful engine.
- max_power: The maximum power output of the car's engine, usually in horsepower (bhp). This indicates how strong the engine is.
- seats: The number of seats the car has, usually 4 or 5 for most cars.
