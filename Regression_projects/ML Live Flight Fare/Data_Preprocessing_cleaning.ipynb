```python
# Price to float
df['Price'] = df['Price'].astype(float)
```

```python
# Date_of_Journey to date time
df['Date_of_Journey'] = pd.to_datetime(df['Date_of_Journey'], format='%d/%m/%Y')
df['Journey_Day'] = pd.to_datetime(df['Date_of_Journey']).dt.day
```

```python
df['Dep_Time'] = pd.to_datetime(df['Dep_Time'], format='%H:%M').dt.time
```

```python
# Ensure 'Dep_Time' is in string format (if not already)
df['Dep_Time'] = df['Dep_Time'].astype(str)

# Split 'Dep_Time' into hours and minutes
df['Dep_Hour'] = df['Dep_Time'].str.split(':').str[0].astype(int)
df['Dep_Minute'] = df['Dep_Time'].str.split(':').str[1].astype(int)

# You can also extract the seconds if needed, assuming no seconds are in the data
df['Dep_Second'] = 0  # Default as 0, since there are no seconds in the 'HH:MM' format

# Verify the result
df[['Dep_Time', 'Dep_Hour', 'Dep_Minute', 'Dep_Second']].head()
```

```python
# Convert 'Arrival_Time' to datetime
df['Arrival_Time'] = pd.to_datetime(df['Arrival_Time'])
```

```python
df['Arrival_Hour'] = pd.to_datetime(df['Arrival_Time']).dt.hour
df['Arrival_Minute'] = pd.to_datetime(df['Arrival_Time']).dt.minute
```

**Duration and Total_Stops columns may require additional processing to extract meaningful numerical data.**

```python
df[['Duration' ,'Total_Stops' ]].head()
```

```python
# duration to timedelta
df['Duration'] = pd.to_timedelta(df['Duration'])


df['Duration_Hour'] = df['Duration'].dt.seconds // 3600
df['Duration_Minute'] = (df['Duration'].dt.seconds % 3600) // 60
```

```python
# Convert 'Total_Stops' to numeric values
df['Total_Stops'] = df['Total_Stops'].replace({'non-stop': 0, '1 stop': 1, '2 stops': 2, '3 stops': 3, '4 stops': 4})

# Verify the result
df[['Total_Stops']].head()
```

**Encoding Categorical Variables**
- Airline, Source, Destination, Route are categorical and need encoding.
- Using Label Encoding or One-Hot Encoding

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Airline'] = le.fit_transform(df['Airline'])
df['Source'] = le.fit_transform(df['Source'])
df['Destination'] = le.fit_transform(df['Destination'])
```
