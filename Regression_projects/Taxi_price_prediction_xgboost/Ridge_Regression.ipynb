# X and y

```python
X = df.drop('Trip_Price', axis=1)
y = df['Trip_Price']
```

# Train Test Split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 5)
```

# Moelling and Evaluation

- finding best alpha value

```python
from sklearn.model_selection import GridSearchCV

# model
from sklearn.linear_model import Ridge
estimator = Ridge()

# parameters and value
param_grid = {"alpha" :  list(range(1,50))}

#Identifying best parameter value within the given values for the given data
model_hp = GridSearchCV(estimator, param_grid, cv = 5, scoring = 'r2')

model_hp.fit(X_train, y_train)
model_hp.best_params_
```

# Modelling Ridge Regression using best hyperparameters

```python
X = df.drop('Trip_Price', axis=1)
y = df['Trip_Price']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2 , random_state = 5)

from sklearn.linear_model import Ridge
ridge_best = Ridge(alpha=17)
ridge_best.fit(X_train, y_train)

print("Intercept : ", ridge_best.intercept_)
print("Coefficient : ", ridge_best.coef_)
print("========================================")

# Prediction and evaluation o train data
ypred_train = ridge_best.predict(X_train)

from sklearn.metrics import r2_score
print("Train_r2 : ", r2_score(y_train, ypred_train))

from sklearn.model_selection import cross_val_score
print("Cross_val_score : ", cross_val_score(ridge_best, X_train, y_train, cv = 5).mean())

# PREDICTION AND EVALUATION ON TEST DATA
ypred_test = ridge_best.predict(X_test)
print("Test_r2 : ", r2_score(y_test, ypred_test))
```
