**X and y**

```python
X = df[['age', 'gender', 'bmi', 'children', 'smoker']]

y = df['charges']
```

# Train test split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state = 9)
```

# Modelling & Evaluation



**Applying hyperparameter tuning for lasso regression**

```python
from sklearn.model_selection import GridSearchCV

# model
from sklearn.linear_model import Lasso
estimator = Lasso()
estimator
```

```python
# parameters and value
param_grid = {"alpha" : list(range(1,100))}


# Identifying best parameter value(alpha) within the given values for the given data
model_hp = GridSearchCV(estimator, param_grid, cv=5, scoring = 'r2')
model_hp
```

```python
model_hp.fit(X_train, y_train)
```

```python
model_hp.best_params_
```

```python
# Write in one cell
from sklearn.model_selection import GridSearchCV

# model
from sklearn.linear_model import Lasso
estimator = Lasso()

# parameters and value
param_grid = {"alpha" : list(range(1,100))}

# Identifying best parameter value(alpha) within the given values for the given data
model_hp = GridSearchCV(estimator, param_grid, cv=5, scoring = 'r2')

model_hp.fit(X_train, y_train)
model_hp.best_params_
```

**BUILD LASSO MODEL USING BEST HYPERPARAMETERS**

```python
from sklearn.linear_model import Lasso
lasso_best = Lasso(alpha=61)
lasso_best.fit(X_train,y_train)

print("Intercept : ", lasso_best.intercept_)
print("coefficients : ", lasso_best.coef_)
print("=============================================")

# Prediction and evaluation on train_data
ypred_train = lasso_best.predict(X_train)

from sklearn.metrics import r2_score
print("Train_R2 : ", r2_score(y_train, ypred_train))

from sklearn.model_selection import cross_val_score
print("Cross_val_score : ", cross_val_score(lasso_best, X_train, y_train, cv = 5).mean())

# Prediction & E valuation on test data
ypred_test = lasso_best.predict(X_test)
print("R2_score : ", r2_score(y_test, ypred_test))
```

# Final Model

```python
X = X.drop("gender", axis = 1)
y = df['charges']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state = 9)


from sklearn.linear_model import Lasso
lasso_best = Lasso(alpha=61)
lasso_best.fit(X_train,y_train)

print("Intercept : ", lasso_best.intercept_)
print("coefficients : ", lasso_best.coef_)
print("=============================================")

# Prediction and evaluation on train_data
ypred_train = lasso_best.predict(X_train)

from sklearn.metrics import r2_score
print("Train_R2 : ", r2_score(y_train, ypred_train))

from sklearn.model_selection import cross_val_score
print("Cross_val_score : ", cross_val_score(lasso_best, X_train, y_train, cv = 5).mean())

# Prediction & E valuation on test data
ypred_test = lasso_best.predict(X_test)
print("R2_score : ", r2_score(y_test, ypred_test))
```
