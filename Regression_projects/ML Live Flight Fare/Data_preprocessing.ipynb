```python
# Use this duration for filling missing values
```

```python
# Calculate total duration in minutes
df['Duration_in_minutes'] = df['Duration_Hour'] * 60 + df['Duration_Minute']

# Loop through rows where 'Route' is missing and fill the values based on closest duration match
for index, row in df[df['Route'].isna()].iterrows():
    source = row['Source']
    destination = row['Destination']
    
    # Filter for flights with the same source and destination
    similar_flights = df[(df['Source'] == source) & (df['Destination'] == destination)]
    
    # Calculate the duration of the current flight with missing values
    missing_flight_duration = row['Duration_Hour'] * 60 + row['Duration_Minute']
    
    # Find the closest match in duration (within a reasonable range)
    closest_match = similar_flights.iloc[(similar_flights['Duration_in_minutes'] - missing_flight_duration).abs().argsort()[:1]]
    
    # Fill the missing Route and Total_Stops from the closest matching flight
    df.at[index, 'Route'] = closest_match['Route'].values[0]
    df.at[index, 'Total_Stops'] = closest_match['Total_Stops'].values[0]

# Check the dataframe after filling
print(df[df['Route'].isna()])
```

```python
# Get the summary statistics (min and max) for the dataframe
desc_stats = df.describe()

# Check for any values less than min or greater than max for each column
out_of_range = {}

for col in continous:  # Check only numerical columns
    min_value = desc_stats.loc['min', col]
    max_value = desc_stats.loc['max', col]
    
    # Find values that are out of the defined range
    out_of_range[col] = {
        "less_than_min": df[col][df[col] < min_value].count(),
        "greater_than_max": df[col][df[col] > max_value].count()
    }

# Display out of range results
out_of_range
```

- **The output shows that for all columns in your dataset, there are no values less than the minimum or greater than the maximum as defined by df.describe(). This means that all the values in each column fall within the expected range.**
